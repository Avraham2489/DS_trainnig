#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
import time

A = np.random.rand(10**7)
B = np.random.rand(10**7)

start_time = time.time()
result_copy = ((A+B)*(-A/2))  
end_time = time.time()
print(f"Time taken with copy: {end_time - start_time:.6f} seconds")


start_time = time.time()
result_in_place = np.empty_like(A)  # יצירת מערך ריק כדי לאחסן את התוצאה
np.negative(A, out=result_in_place)  # חישוב -A במקומו
result_in_place /= 2  # חיבור ב-2 במקומו
result_in_place *= (A + B)  # הכפל הסופי במקומו
end_time = time.time()
print(f"Time taken in place: {end_time - start_time:.6f} seconds")

````

#### 56. Generate a generic 2D Gaussian-like array (★★☆)

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

A = 1              
mu_x, mu_y = 0, 0 
sigma_x, sigma_y = 1, 1 

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Compute the 2D Gaussian function
G = A * np.exp(-((X - mu_x)**2 / (2 * sigma_x**2) + (Y - mu_y)**2 / (2 * sigma_y**2)))

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, G, cmap='viridis')
plt.show()
````

#### 63. Create an array class that has a name attribute (★★☆)

````python

class NamedArray:
    def __init__(self, name, data):
        self.name = name
        self.data = np.array(data)

    def __repr__(self):
        return f"NamedArray(name='{self.name}', data={self.data.tolist()})"

    def __str__(self):
        return f"{self.name}: {self.data}"

    def __getitem__(self, index):
        return self.data[index]

    def __setitem__(self, index, value):
        self.data[index] = value

    def __len__(self):
        return len(self.data)

# Example usage
if __name__ == "__main__":
    arr = NamedArray("My Array", [1, 2, 3, 4, 5])
    print(arr)
````

#### 90. Given an arbitrary number of vectors, build the cartesian product (every combinations of every item) (★★★)

````python
def cartesian(arrays):
    arrays = [np.asarray(a) for a in arrays]
    shape = (len(x) for x in arrays)
    ix = np.indices(shape, dtype=int)
    ix = ix.reshape(len(arrays), -1)
    ix=ix.T
    for n, y in enumerate(arrays):
        ix[:, n] = arrays[n][ix[:, n]]
    return ix

print (cartesian(([4, 5], [6, 7])))
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?

````python


def sample_equidistant(X, Y, num_samples):
    dX = np.diff(X)
    dY = np.diff(Y)

    distances = np.sqrt(dX**2 + dY**2)

    cumulative_distances = np.concatenate([[0], np.cumsum(distances)])

    total_length = cumulative_distances[-1]
    equidistant_points = np.linspace(0, total_length, num_samples)

    X_equidistant = np.interp(equidistant_points, cumulative_distances, X)
    Y_equidistant = np.interp(equidistant_points, cumulative_distances, Y)

    return X_equidistant, Y_equidistant

X = np.array([0, 1, 4, 6])
Y = np.array([0, 1, 0, 3])

num_samples = 10

X_sampled, Y_sampled = sample_equidistant(X, Y, num_samples)

print(X_sampled)
print(Y_sampled)
````

Example
Let us assume the following comma separated input sequence is given to the program:

2. At least 1 number between [0-9]
1. At least 1 letter between [A-Z]
3. At least 1 character from [$#@]

Level 3

Question:
Define a class with a generator which can iterate the numbers, which are divisible by 7, between a given range 0 and n.

Hints:
Consider use yield



```python
class DivisibleBySeven:
    def __init__(self, n):
        self.n = n

    def generate(self):
        for num in range(0, self.n + 1):
            if num % 7 == 0:
                yield num 

n = int(input("Enter a number: "))

generator = DivisibleBySeven(n)

print(",".join(map(str, generator.generate())))
    
```

UP 5

¡­
The numbers after the direction are steps. Please write a program to compute the distance from current position after a sequence of movement and original point. If the distance is a float, then just print the nearest integer.
Example:

UP 5
DOWN 3

```python
from math import sqrt

y_sum=0
X_sum=0

while True:
    input1=input("tup")
    if not input1:  
        break 

    dirracion,sum_total = input1.split(" ")

    if dirracion == "UP":
        y_sum+=int(sum_total)
    elif dirracion == "DOWN":
        y_sum-=int(sum_total)
    elif dirracion == "LEFT":
        X_sum+=int(sum_total)
    elif dirracion == "RIGHT":
        X_sum-=int(sum_total)

print(round(sqrt((X_sum)**2+(y_sum)**2)))
                                
```


Question:
Please write a program using generator to print the numbers which can be divisible by 5 and 7 between 0 and n in comma separated form while n is input by console.
Example:
If the following n is given as input to the program:
100
Then, the output of the program should be:
0,35,70

Hints:
Use yield to produce the next value in generator.
In case of input data being supplied to the question, it should be assumed to be a console input.



```python
from math import gcd

lcm = (5 * 7) // gcd(5, 7)

def divisible_by_5_and_7(n):
    for i in range(0, n + 1, lcm):
        yield i

n = int(input("Enter numbers: "))

print(",".join(map(str, divisible_by_5_and_7(n))))
```


Question:
Please write a binary search function which searches an item in a sorted list. The function should return the index of element to be searched in the list.


Hints:
Use if/elif to deal with conditions.




```python
def bin_s(list1, goal):
    left = 0
    right = len(list1) - 1

    while left <= right:  
        mid = (left + right) // 2  

        if list1[mid] == goal:  
            return mid
        elif list1[mid] < goal:  
            left = mid + 1  
        else:  
            right = mid - 1  

    return -1  

x = [1, 2, 2, 3, 4, 6, 8, 9, 14, 15, 18]
goal = 6
print(bin_s(x, goal))
```
